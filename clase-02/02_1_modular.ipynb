## Concepto de modularización y encapsulado

Una característica importante de la programación es la posibilidad de definir partes del código para realizar alguna tarea específica y poder reutilizarla. En Python, una de las maneras más efectivas de hacer esto es mediante la definición de *funciones*. Ya vimos un ejemplo, con la función `print()`, que se usa para imprimir valores a pantalla (por defecto) o a un archivo. Existen otras funciones ya definidas en Python (`len()`, `max()`, `min()`, `abs()`, etc). Puede ver la [lista completa en la documentación](https://docs.python.org/3/library/functions.html).

## Continuación de nuestro problema

Definimos una función muy simple para calcular el área de un círculo de diámetro $d$, siguiendo la ecuación:

$$ A = \pi \times \left(\frac{d}{2}\right)^2$$

Incluso en este caso sencillo aparece un problema: El cálculo va a tener un error asociado al valor aproximado de $\pi \approx 3.1416$ que estamos usando.

La constante $\pi$, al igual que las funciones matemáticas más elementales y otras más *sofisticadas* no son parte del lenguaje.

Esta es una situación que vamos a encontrar repetidamente en cualquier aplicación del lenguaje, ya que es relativamente pequeño.

Sin embargo, existe una biblioteca con paquetes con facilidades para muchas áreas útiles de programación. Estas facilidades no son parte intrínseca del lenguaje pero vienen incluidas y son de muy fácil acceso. 


## Paquetes y Módulos

Un paquete es un módulo o un conjunto de módulos relacionados temáticamente.

Los módulos son el mecanismo de Python para reusar código. Existen varios módulos que son parte de la biblioteca *standard* para distintos tipos de aplicaciones. Además existe un repositorio en la que contribuye la comunidad, con bibliotecas para muchos usos diversos.

El uso de módulos es muy simple, para poder aprovecharlo necesitaremos saber dos cosas:

* Qué funciones están ya definidas y listas para usar
* Cómo acceder a ellas

Empecemos con la segunda cuestión. Para utilizar las funciones debemos *importarlas* en la forma `import modulo`, donde modulo es el nombre que queremos importar.

Esto nos lleva a la primera cuestión: cómo saber ese nombre, y que funciones están disponibles. La respuesta es: **la documentación** de python o algún tipo de investigación previa.

Una vez importado, podemos utilizar constantes y funciones definidas en el módulo con la notación "de punto": `modulo.consante` o `modulo.funcion()`.

### Ejemplo: módulo math

El módulo **math** contiene las funciones más comunes (trigonométricas, exponenciales, logaritmos, etc) para operar sobre números de *punto flotante*, y algunas constantes importantes ($\pi$, e, etc). En realidad es una interface a la biblioteca math en C.

```python
import math
# algunas constantes y funciones elementales
print ("e =",math.e, ",  π =", math.pi)

print (math.sqrt(5 * math.pi))
# otras
print ('Conversión radianes -> grados: π/4 =',math.degrees(math.pi/4),"°")
print ('Conversión grados -> radianes: 45° =',math.radians(45))
```

```python
help(math)
```

A veces, sólo necesitamos unas pocas funciones de un módulo. Entonces para abreviar la notación combiene importar sólo lo que vamos a usar, usando la notación:

   `from modulo import func1, func2, const1`

```python
from math import sqrt, pi, log
raiz5pi = sqrt(5*pi)
print (log(1024, 2))
```

Otra manera muy popular de importar un módulo es darle un *alias*:

```python
import math as m
m.sqrt(5*m.pi)
```

En nuestro ejemplo, podemos obtener muy fácilmente no sólo una solución más precisa sino más consistente utilizando el valor de $\pi$ ya definido en el módulo `math`:

```python
from math import pi
def area_circulo(diam):
    """Cálculo de la superficie de un círculo de diámetro diam."""
    r = diam/2  # radio

    # Cálculo
    A = pi * r**2    # Área del círculo
    return A
```
