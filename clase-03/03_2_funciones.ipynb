En **Python** `function` es un objeto, con una operación importante: podemos llamarla (ejecutarla), en la forma: `func(lista-de-argumentos)`

Como con todos los objetos, podemos definir una variable y asignarle una función (algo así como lo que en C sería un puntero a funciones)

En Python, las funciones son de _primera clase_, es decir que tienen las siguientes propiedades:

- Asignación a Variables: Se puede asignar la función a una variable, lo cual permite referenciarla de manera similar a como se haría con un número, cadena de texto u otro tipo de dato. Esto ocurre en Python dado que todo elemento del lenguaje es un objeto.

- Pasaje como argumento: Se puede pasar la función como argumento a otras funciones.

- Retorno de Funciones: Una función puede devolver otra función como valor de retorno

- Almacenamiento en Estructuras de Datos: Se pueden almacenar funciones en estructuras de datos como listas, diccionarios, y otros contenedores.


## Definición básica de funciones

Ya vimos en la segunda clase como definir una función.


Consideremos una función para construir un correo electrónico (*email*):

````python
def format_email(to, subject, cc, body, signature):
    """
    Crea un email simple.
    Argumentos
    ----------
      to (str): recipiente
      subject (str): asunto del email
      cc (list de str): CC recipientes
      body (str): cuerpo del texto
      signature (str): despedida

    Returns
    -------
      email (str): el email creado
    """
    cc_part = f"Cc: {', '.join(cc)}\n"
    email = f"To: {to}\n" + f"{cc_part}Subject: {subject}\n{body}\n\n{signature}"
    return email
````

````python
help(format_email)
````

````python
a_quien = "usted@example.com"
ccs = ["ana@example.com", "flavio@example.com"]
cuerpo = "Envío adjunta la lista de notas del curso."
email = format_email(a_quien, "Notas curso", ccs, cuerpo, "Saludos,\n\tJuan")
````

````python
print(email)
````

Notemos que podemos llamar a esta función de varias maneras. Podemos llamarla con la constante, o con una variable indistintamente. En este caso, cada argumento está definido por su posición.

Pero en Python podemos usar el nombre de la variable en el llamado a la función. 

En ese caso no es necesario respetar el orden de los argumentos. Por ejemplo:

````python
print(format_email(a_quien, "Notas", ccs, signature="Saludos,\n\tJuan", body=cuerpo))
````


## Funciones con argumentos opcionales

Las funciones pueden tener muchos argumentos. En **Python** pueden tener un número variable de argumentos y pueden tener valores por *default* para algunos de ellos. En el caso de la función anterior, vamos a extenderlo de manera que podamos usarlo de manera sencilla:

````python
def format_email(to, subject, cc, body=None, signature="Saludos,\n\tJuan"):
    """
    Crea un email simple.
    Argumentos
    ----------
      to (str): recipiente
      subject (str): email subject
      cc (list of str): CC recipientes
      body (str): cuerpo del texto (opcional)
      signature (str): despedida (opcional)

    Returns
    -------
      email (str): el email creado
    """
    cc_part = f"Cc: {', '.join(cc)}\n"
    if body is None:
        cuerpo = "Sin texto\n"
    else:
        cuerpo = body

    email = f"To: {to}\n" + f"{cc_part}Subject: {subject}\n{cuerpo}\n\n{signature}"

    return email
````

````python
print(format_email(a_quien, "Notas", ccs, body=cuerpo))
````

Si mantenemos el orden no es necesario especificar el nombre de los argumentos, aunque sean opcionales:

````python
print(format_email(a_quien, "Notas", ccs, cuerpo))
````

````python
help(format_email)
````

````python
print(format_email(a_quien, "Notas", ccs, signature="Chau"))
````

````python
print(format_email(a_quien, "Notas", ccs, "Chau"))  # resultado inesperado
````


## Número variable de argumentos y argumentos *keywords* 

Se pueden definir funciones que toman un número variable de argumentos (como una lista), o que aceptan un diccionario como argumento. Este tipo de argumentos se llaman argumentos *keyword* (``kwargs``). 
Una buena explicación se encuentra en el [Tutorial de la documentación](https://docs.python.org/3/tutorial/controlflow.html#keyword-arguments). Ahora vamos a dar una explicación rápida. Consideremos la función `f`, que imprime sus argumentos:

````python
def f(p, *args, **kwargs):
  print( f"p     : {p}, tipo: {type(p)}")
  print( f"args  : {args}, tipo: {type(args)}")
  print( f"kwargs: {kwargs}, tipo: {type(kwargs)}")
````

````python
f(1)
````

````python
f(1,2,3)
````

````python
f(1,2,3,4,5,6)
````

En este ejemplo, el primer valor se asigna al argumento requerido `p`, y los siguientes a una variable que se llama `args`, que es del tipo `tuple`

````python
f(1.5, 2, 3, 5, anteultimo = 9, ultimo = -1)
````

En estas otras llamadas a la función, todos los argumentos que se pasan indicando el nombre se asignan a un diccionario.

Al definir una función, con la construcción `*args` se indica *"mapear todos los argumentos posicionales no explícitos a una tupla llamada `args`"*. Con `**kwargs` se indica "mapear todos los argumentos de palabra clave no explícitos a un diccionario llamado `kwargs`". Esta acción de convertir un conjunto de argumentos a una tuple o diccionario se conoce como *empacar* o *empaquetar* los datos.

------

>**NOTA:** Por supuesto, no es necesario utilizar los nombres "args" y "kwargs". Podemos llamarlas de cualquier otra manera! los simbolos que indican cantidades arbitrarias de parametros son `*` y `**`. Además es posible poner parametros "comunes" antes de los parametros arbitrarios, como se muestra en el ejemplo anterior.

------


Un ejemplo de una función con número variable de argumentos puede ser la función `multiplica`:

````python
def multiplica(*args):
  s = 1
  for a in args:
    s *= a
  return s
````

````python
multiplica(2,5)
````

````python
multiplica(2,3,5,9,4)
````

````python
multiplica()
````

## Empacar y desempacar argumentos


Cuando en **Python** creamos una función que acepta un número arbitrario de argumentos estamos utilizando una habilidad del lenguaje que es el "empaquetamiento" y "desempaquetamiento" automático de variables.

Al definir un número variable de argumentos de la forma:
```python
def f(*v):
...
```

y luego utilizarla en alguna de las formas:

```python
f(1)
f(1,'hola')
f(a,2,3.5, 'hola')
```

**Python** automáticamente convierte los argumentos en una única tupla:
```python
f(1)                 --> v = (1,)
f(1,'hola')          --> v = (1,'hola')
f(a,2,3.5, 'hola')   --> v = (a,2,3.5,'hola')
```

Análogamente, cuando utilizamos funciones podemos desempacar múltiples valores en los argumentos de llamada a las funciones.

Si definimos una función que recibe un número determinado de argumentos
```python
def g(a, b, c):
...
```
y definimos una lista (o tupla)
```python
t1 = [a1, b1, c1]
```
podemos realizar la llamada a la función utilizando la notación "asterisco" o "estrella"
```python
g(*t1)             -->  g(a1, b1, c1)
```

Por ejemplo, podemos ejecutar la función definida anteriormente de la siguiente manera:

````python
argumentos = [a_quien, "Notas", ccs, cuerpo, "chau"]
````

````python
print(type(argumentos))
````

````python
print(format_email(*argumentos))   # equivalente a: format_email(a_quien, "Notas", ccs, cuerpo, "chau")
````

En la llamada a la función, la expresión `*argumentos` le indica al intérprete Python que la secuencia (lista o tuple) debe convertirse en una sucesión de argumentos, que es lo que acepta la función. 

> **Nota:** Esta notación se puede utilizar en el llamado de funciones pero **NO** se puede utilizar en cualquier contexto.
>
> Por ejemplo, es un error intentar:
>
> ```python
>t2 = *t1
>```



Similarmente, para desempacar un diccionario usamos la notación `**diccionario`:

````python
# diccionario, caída libre en la luna
otros_args = {'to':a_quien, 'cc': ccs, "body" : "Envío las notas", 
              "subject":'Notas del curso', "signature": "\tEsto es todo, amigos!"} 

correo = format_email(**otros_args)
print (f"Este el el correo:\n{18*'-'}\n{correo}")
````

**En resumen:**

* la notación `(*argumentos)` convierte la lista (o tupla) en los tres argumentos que acepta la función. Los siguientes llamados son equivalentes:
```python
format_email(*argumentos)
format_email(argumentos[0], argumentos[1], argumentos[2], argumentos[3], arguments[4])
format_email(a_quien, "Notas", ccs, cuerpo, signature)
```
* la notación `(**otros_args)` desempaca el diccionario en pares `clave=valor`, siendo equivalentes los dos llamados:
```python
format_email(**otros_args)
format_email(to=a_quien, cc=ccs, body="Envío las notas",subject="Notas del curso", signature="\tEsto es todo, amigos!")
```



## Funciones que toman como argumento una función

````python
def aplicar_fun(f, L):
  """Aplica la función f a cada elemento del iterable L 
  devuelve una lista con los resultados.
  
  IMPORTANTE: Notar que no se realiza ninguna comprobación de validez
  """
  return [f(x) for x in L]
````

> En lenguajes funcionales la función `aplicar_fun` se llama _map_

````python
import math as m
degs = list(range(0,190,20))
angles = aplicar_fun(m.radians, degs)
````

````python
degs
````

````python
aplicar_fun(m.sin, angles)
````

Notar que definimos la función `aplicar_fun()` que recibe una función y una secuencia, pero no necesariamente una lista, por lo que podríamos aplicarla directamente a `range`. Quizás queda un poco más comprimido pero podríamos juntar las líneas anteriores a:

````python
aplicar_fun(m.sin, aplicar_fun(m.radians, range(0,190,20)))
````

Además, debido a su definición, el primer argumento de la función `aplicar_fun()` no está restringida a funciones numéricas pero al usarla tenemos que asegurar que la función y el iterable (lista) que pasamos como argumentos son compatibles.

Veamos otro ejemplo:

````python
s = ['hola', 'chau']
print(aplicar_fun(str.upper, s))
````

donde `str.upper` es una función definida en python, que convierte a mayúsculas el string dado `str.upper('hola') = 'HOLA'`.

## Funciones que devuelven una función

Al ser las funciones objetos, en igualdad de condiciones que otros tipos, pueden ser retornadas por una función como cualquier otro objeto (números, listas, tuples, cadenas de caracteres, diccionarios, etc). Veamos un ejemplo simple de funciones que devuelven una función:

````python
def make_poli(coef):
    """
    Crea una función polinómica a partir de sus coeficientes
    Argumentos
    ----------
      coef (list or tuple of numbers): coeficientes del polinomio
    
    Returns
    -------
      Polinomio (func): el polinomio coef[0] + coef[1]*x + coef[2]*x**2 + ... coef[n]*x**n
    """
    n = len(coef)  # Grado del polinomio 
    def poli(x):
        return sum([coef[j]*x**j for j in range(n)])
    return poli
````

````python
cubico = make_poli([0,0,0,1])  # f(x) = x³
````

````python
cubico(2.)
````

````python
type(cubico)
````

Incluso podemos combinar las dos funciones definidas anteriormente

````python
aplicar_fun(make_poli([1,1,1]), [1,2,3])  # x^2 + x + 1 aplicado a 1, 2, y 3
````


## Funciones anónimas

En algunas ocasiones suele ser conveniente definir una función que se va a utilizar una única vez sin darle un nombre. Estas se llaman funciones *lambda*, y su uso más común es como argumento de otras funciones. Consideremos el caso en que queremos ordenar alfabéticamente una lista de *strings*:

````python
s1 = ['Estudiantes', 'caballeros', 'Python', 'Curso', 'pc', 'azufre']
print(s1)
print(sorted(s1))
````

Como los elementos son *strings*, la comparación se hace respecto a su posición en el abecedario. En este caso no es lo mismo mayúsculas o minúsculas, y el resultado no es el esperado.
Si queremos el orden independiente de mayúsculas o minúsculas podemos usar el argumento `key` de la función `sorted`:

````python
sorted(s1, key=str.lower)
````

Aquí utilizamos la función `str.lower` pero también podemos definir una función propia.
Por ejemplo, si queremos ordenar por la segunda letra, definimos una función que devuelva este caracter:

````python
def segunda(s):
    return s[1]
````

````python
sorted(s1, key=segunda)
````

````python
sorted(s1, key=lambda a: a[1])
````

Similarmente, si queremos ordenarla alfabéticamente empezando desde atrás hacia adelante:

````python
sorted(s1, key=lambda a: str.lower(a[::-1]))
````

>**Nota:** La forma general de las funciones `lambda` es:
>
>```python
>lambda x,y,z: expresión_de(x,y,z)
>```
